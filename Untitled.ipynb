```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('seaborn-v0_8-darkgrid')

fig, ax = plt.subplots(figsize=(12, 7))

n, bins, patches = ax.hist(df['popularity'], 
                           bins=20, 
                           color='steelblue', 
                           edgecolor='black', 
                           alpha=0.8,
                           linewidth=1.2,
                           rwidth=0.95)

ax.set_title('Распределение популярности треков (popularity)', 
             fontsize=16, 
             fontweight='bold',
             pad=20)

ax.set_xlabel('Значение popularity', fontsize=14, fontweight='medium')
ax.set_ylabel('Количество треков', fontsize=14, fontweight='medium')

ax.set_xlim(0, 100)
ax.grid(True, alpha=0.3)

median_val = df['popularity'].median()
ax.axvline(median_val, color='red', linestyle='--', 
           linewidth=2, label=f'Медиана: {median_val:.0f}')

mean_val = df['popularity'].mean()
ax.axvline(mean_val, color='green', linestyle='--', 
           linewidth=2, label=f'Среднее: {mean_val:.1f}')

ax.legend(fontsize=12, loc='upper right')

stats_text = f'Всего треков: {len(df)}\nСреднее: {mean_val:.1f}\nМедиана: {median_val:.0f}\nМин: {df["popularity"].min()}\nМакс: {df["popularity"].max()}'
ax.text(0.02, 0.95, stats_text, transform=ax.transAxes, 
        fontsize=12, verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.tight_layout()
plt.show()


print("АНАЛИЗ ГИСТОГРАММЫ РАСПРЕДЕЛЕНИЯ POPULARITY:")


print("\n1. ОСНОВНАЯ ЧАСТЬ ЗНАЧЕНИЙ НАХОДИТСЯ В ДИАПАЗОНЕ:")

percentiles = np.percentile(df['popularity'], [25, 50, 75])
print(f"   25-й процентиль (Q1): {percentiles[0]:.0f}")
print(f"   50-й процентиль (медиана): {percentiles[1]:.0f}")
print(f"   75-й процентиль (Q3): {percentiles[2]:.0f}")
print(f"   Основная часть значений (межквартильный размах): от {percentiles[0]:.0f} до {percentiles[2]:.0f}")
print(f"   50% всех треков имеют popularity в диапазоне [{percentiles[0]:.0f}, {percentiles[2]:.0f}]")
print(f"   Пик распределения приходится на диапазон 20-40")

print("\n2. ЗНАЧЕНИЯ, КОТОРЫЕ ВСТРЕЧАЮТСЯ ЗНАЧИТЕЛЬНО РЕЖЕ ОСТАЛЬНЫХ:")

q1 = df['popularity'].quantile(0.25)
q3 = df['popularity'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers_low = df[df['popularity'] < lower_bound]
outliers_high = df[df['popularity'] > upper_bound]

print(f"   Нижняя граница для выбросов: {lower_bound:.0f}")
print(f"   Верхняя граница для выбросов: {upper_bound:.0f}")
print(f"   Количество треков с низкой популярностью (выбросы слева): {len(outliers_low)}")
print(f"   Количество треков с высокой популярностью (выбросы справа): {len(outliers_high)}")

if len(outliers_high) > 0:
    print(f"\n   Примеры треков с АНОМАЛЬНО ВЫСОКОЙ популярностью (> {upper_bound:.0f}):")
    print(outliers_high[['track_name', 'artist_name', 'popularity']].head(5).to_string(index=False))

if len(outliers_low) > 0:
    print(f"\n   Примеры треков с АНОМАЛЬНО НИЗКОЙ популярностью (< {lower_bound:.0f}):")
    print(outliers_low[['track_name', 'artist_name', 'popularity']].head(5).to_string(index=False))

print(f"\n   ВЫВОД: Значения popularity распределены неравномерно.")
print(f"   - Большинство треков (около 75%) имеют popularity в диапазоне 20-50")
print(f"   - Треков с очень высокой популярностью (> 70) значительно меньше")
print(f"   - Треков с очень низкой популярностью (< 10) тоже мало")
print(f"   - Распределение смещено влево (в сторону низких значений)")
```

Что такое DataFrame?  
DataFrame — это двумерная табличная структура данных в библиотеке Pandas, содержащая строки и столбцы. Каждый столбец может содержать данные разного типа, а строки идентифицируются индексами.
Что такое признак (Feature)? 
Признак — это отдельный столбец в наборе данных, который описывает какую-то характеристику объекта. Например, в датасете о треках признаками являются: `popularity`, `genre`, `stream_count` и другие.
Чем среднее арифметическое отличается от медианы?
Среднее арифметическое — это сумма всех значений, делённая на их количество. Оно чувствительно к выбросам.  
Медиана — это значение, которое делит упорядоченный ряд данных пополам. Она устойчива к выбросам и лучше отражает «типичное» значение в асимметричных распределениях.
Что показывает гистограмма?
Гистограмма показывает распределение числовых данных: на каких значениях сосредоточено большинство наблюдений, есть ли выбросы, симметрично ли распределение и какова его форма (например, нормальное или скошенное).
Для чего нужна фильтрация данных?
Фильтрация позволяет отобрать только те строки, которые удовлетворяют заданному условию. Это нужно для:
удаления шума и пропусков,
анализа конкретных подгрупп данных,
сравнения разных категорий,
подготовки данных для дальнейшего анализа или машинного обучения.
